### Ejercicios de Funciones

#### 1.- El menor de dos pares. Escribe una función que devuelva el menor de dos números, SOLAMENTE si los dos números son pares. En caso de que uno de los dos, o los dos, sean impares, devuelve el número más alto

```python

def menor_o_mayor_pares(a, b):
    # Verificamos si ambos números son pares (su residuo al dividir entre 2 es 0)
    if a % 2 == 0 and b % 2 == 0:
        # Si los dos son pares, devolvemos el menor
        return min(a, b)
    else:
        # Si uno de los dos (o ambos) es impar, devolvemos el más alto
        return max(a, b)

# --- Ejemplos de prueba ---
print(menor_o_mayor_pares(4, 8))   # Ambos son pares -> Devuelve el menor: 4
print(menor_o_mayor_pares(4, 7))   # Uno es impar -> Devuelve el mayor: 7
print(menor_o_mayor_pares(3, 5))   # Ambos son impares -> Devuelve el mayor: 5
```

#### 2.- Escribe una función que tome como argumento una string de dos palabras y devuelva True si ambas empiezan por la misma letra

```python
def empiezan_misma_letra(texto):
    # Separamos el string en una lista con sus palabras
    palabras = texto.split()
    
    # Comparamos la primera letra de la primera palabra (palabras[0][0]) 
    # con la primera letra de la segunda palabra (palabras[1][0]).
    # Usamos .lower() para que la comparación no distinga entre mayúsculas y minúsculas.
    return palabras[0][0].lower() == palabras[1][0].lower()

# --- Ejemplos de prueba ---
print(empiezan_misma_letra("Gato grande"))  # Devuelve True (G y g)
print(empiezan_misma_letra("Perro lobo"))   # Devuelve False (P y l)
print(empiezan_misma_letra("Casa caro"))    # Devuelve True (C y c)
```

#### 3.- Escribe una función que tome dos números enteros y que devuelva True si la suma de los dos es 20, o si uno de los dos es 20. Que devuelva Falso si no.

```python
def es_veinte_o_suma(a, b):
    # Comprobamos si a es 20, b es 20, o si la suma de ambos es 20
    if a == 20 or b == 20 or (a + b) == 20:
        return True
    else:
        return False

# --- Ejemplos de prueba ---
print(es_veinte_o_suma(10, 10))  # Su suma es 20 -> Devuelve True
print(es_veinte_o_suma(20, 5))   # Uno de ellos es 20 -> Devuelve True
print(es_veinte_o_suma(7, 13))   # Su suma es 20 -> Devuelve True
print(es_veinte_o_suma(5, 5))    # Ninguna condición se cumple -> Devuelve False
```

#### 4.- Escribe una función que devuelva en mayúsculas la primera letra y la quinta letra de una palabra. Pista: usa capitalize()

```python
def primera_y_quinta(palabra):
    # Verificamos que la palabra tenga al menos 5 letras para evitar errores de índice
    if len(palabra) < 5:
        return "La palabra debe tener al menos 5 letras."
    
    # 1. Usamos capitalize() para poner la primera letra en mayúscula (y el resto en minúscula)
    primera_mayus = palabra.capitalize()
    
    # 2. Reconstruimos la palabra manteniendo el resto, pero asegurándonos de que la 5ª letra (índice 4) sea mayúscula
    resultado = primera_mayus[:4] + palabra[4].upper() + primera_mayus[5:]
    
    return resultado

# --- Ejemplos de prueba ---
print(primera_y_quinta("programacion"))  # Devuelve: ProgrAmacion
print(primera_y_quinta("computadora"))   # Devuelve: CompUtadora
```

#### 5.- Escribe una función que devuelva una frase invertida.

```python
def invertir_frase(frase):
    # El truco [::-1] recorre el string completo en dirección inversa
    return frase[::-1]

# --- Ejemplos de prueba ---
print(invertir_frase("Hola mundo"))  # Devuelve: odnum aloH
print(invertir_frase("Python"))     # Devuelve: nohtyP
```

#### 6.- Dada una lista de enteros, devuelve True si contiene un 3 al lado de otro 3

```python
def tiene_tres_dobles(lista):
    # Recorremos la lista hasta el penúltimo elemento para no salirnos del rango
    for i in range(len(lista) - 1):
        # Comprobamos si el número actual y el siguiente son 3
        if lista[i] == 3 and lista[i+1] == 3:
            return True
    return False

# --- Ejemplos de prueba ---
print(tiene_tres_dobles([1, 3, 3, 4]))  # Devuelve True (hay dos 3 seguidos)
print(tiene_tres_dobles([3, 1, 3, 3]))  # Devuelve True (al final hay dos 3 seguidos)
print(tiene_tres_dobles([1, 3, 1, 3]))  # Devuelve False (los 3 están separados)
```

#### 7.- Escribe una función que busque numeros primos. 

```python
def es_primo(n):
    # Los números menores de 2 no son primos
    if n < 2:
        return False
    
    # Comprobamos si es divisible por algún número desde el 2 hasta su raíz cuadrada
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False  # Encontramos un divisor, por lo que no es primo
            
    return True  # Si no encontró divisores, es primo

# --- Ejemplos de prueba ---
print(es_primo(7))   # Devuelve True
print(es_primo(10))  # Devuelve False
print(es_primo(13))  # Devuelve True
```

#### 8.- Juego de espías. Escribe una función que encuentre la secuencia 007 en una lista

```python
def juego_espias():
    print("--- BIENVENIDO AL JUEGO DE ESPÍAS ---")
    print("Introduce una lista de números separados por espacios (ej: 1 2 0 0 7 5):")
    
    # Capturamos la entrada del usuario por teclado (como en el juego de la "O")
    entrada = input("Tus números: ")
    
    # Convertimos el texto ingresado en una lista de números enteros
    lista = [int(x) for x in entrada.split()]
    
    # Lógica para encontrar la secuencia [0, 0, 7]
    codigo = [0, 0, 7]
    
    for num in lista:
        if len(codigo) > 0 and num == codigo[0]:
            codigo.pop(0)
            
    # Comprobamos el resultado e imprimimos un mensaje interactivo
    if len(codigo) == 0:
        print("¡Misión cumplida! Secuencia 007 encontrada con éxito.")
        return True
    else:
        print("¡Fallaste! El agente secreto no dejó rastro (falta la secuencia 007).")
        print("Tu lista analizada era:", lista)
        return False

# --- Para jugar, simplemente ejecuta esta línea: ---
juego_espias()
```

#### 9.- Juego de BLACKJACK. Dados tres numeros enteros entre el 1 y el 11, si su suma es menor o igual que 21, devuelve su suma. Si su suma es mayor que 21 y hay un 11, resta 10 al total de la suma. Finalmente, si con el ajuste es mayor que 21, se devuelve BUST.
##### 5,6,7 =18
##### 9,9,9 = BUST
##### 9,9,11 = 19

```python
def jugar_blackjack():
    print("--- 🃏 BIENVENIDO AL JUEGO DE BLACKJACK 🃏 ---")
    print("Introduce tres números enteros entre 1 y 11 separados por espacios (ejemplo: 9 9 11):")
    
    # Capturamos la entrada del usuario de forma interactiva (estilo clase)
    entrada = input("Tus tres números/cartas: ")
    
    # Convertimos el texto introducido en una lista de números enteros
    cartas = [int(x) for x in entrada.split()]
    
    # Validamos que se hayan introducido exactamente tres números
    if len(cartas) != 3:
        print("⚠️ ¡Atención! Debes introducir exactamente tres números separados por un espacio.")
        return
    
    a, b, c = cartas[0], cartas[1], cartas[2]
    suma = a + b + c
    
    print(f"\nHas elegido: {a}, {b}, {c}")
    print(f"Suma inicial: {suma}")
    
    # 1. Si la suma es menor o igual a 21
    if suma <= 21:
        print(f"✅ ¡Resultado ganador! La suma es menor o igual a 21: {suma}")
        return suma
    
    # 2. Si es mayor a 21 y hay un 11, aplicamos el ajuste (-10)
    elif (a == 11 or b == 11 or c == 11) and (suma - 10 <= 21):
        suma_ajustada = suma - 10
        print(f"🔄 ¡Había un 11! Se ajusta restando 10. Tu puntuación final es: {suma_ajustada}")
        return suma_ajustada
    
    # 3. Si aun con el ajuste (o sin el 11) supera 21, devolvemos BUST
    else:
        print("❌ ¡BUST! Te has pasado de 21.")
        return "BUST"

# --- Ejecuta esta línea para probar el juego interactivo ---
jugar_blackjack()
```
