#### 2.- Dada la tupla _t_, convierte la tupla en un set. Calcula la suma de todos los elementos únicos y convierte el set de nuevo en tupla ordenada de mayor a menor.

```python
t = (4, 8, 15, 16, 23, 45, 6, 7, 9, 5, 111)
```

```python
# 1. Convertir la tupla en un set
mi_set = set(t)

# 2. Calcular la suma de todos los elementos únicos
suma_total = sum(mi_set)

# 3. Convertir el set de nuevo en una tupla ordenada de mayor a menor
# Usamos sorted() con reverse=True y luego lo convertimos a tuple()
tupla_ordenada = tuple(sorted(mi_set, reverse=True))

# Mostramos los resultados
print("Set creado:", mi_set)
print("Suma de los elementos únicos:", suma_total)
print("Tupla ordenada de mayor a menor:", tupla_ordenada)
```

#### 3.- Dada la cadena _texto_, crea un diccionario donde la clave sea cada letra y el valor sea la cantidad de veces que aparece. Convierte el diccionario en una lista de tuplas (letra, frecuencia) ordenada de mayor a menor frecuencia. 

```python
texto = "banana"
```

```python
# 1. Crear el diccionario donde la clave es la letra y el valor su frecuencia
frecuencias = {}
for letra in texto:
    if letra in frecuencias:
        frecuencias[letra] += 1
    else:
        frecuencias[letra] = 1

# 2. Convertir el diccionario en una lista de tuplas (letra, frecuencia) ordenada de mayor a menor
# Usamos .items() para obtener los pares y sorted() con una función lambda basada en el valor (x[1])
lista_ordenada = sorted(frecuencias.items(), key=lambda x: x[1], reverse=True)

# Mostramos los resultados
print("Diccionario de frecuencias:", frecuencias)
print("Lista de tuplas ordenada de mayor a menor:", lista_ordenada)
```

#### 6.- Dado el string _frase_, obtén un set con todas las palabras únicas. Crea un diccionario que cuente cuántas veces aparece cada palabra. Devuelve el diccionario pero solo con las palabras que aparezcan más de 1 vez.

```python
frase = "En el camino, uno se da cuenta de que en un coche, uno siempre está en un compartimento, y, porque está acostumbrado, no se da cuenta de que, a través de esa ventana, todo lo que ve es simplemente más televisión"
```

```python
# Limpiamos el texto: pasamos a minúsculas y quitamos comas y puntos para evitar errores de conteo
frase_limpia = frase.lower()
for signo in ",.":
    frase_limpia = frase_limpia.replace(signo, "")

# Separamos la frase en una lista de palabras usando los espacios
palabras = frase_limpia.split()

# 1. Obtener un set con todas las palabras únicas
palabras_unicas = set(palabras)

# 2. Crear un diccionario que cuente cuántas veces aparece cada palabra
conteo_palabras = {}
for palabra in palabras:
    # .get() busca la clave y si no existe devuelve 0 por defecto
    conteo_palabras[palabra] = conteo_palabras.get(palabra, 0) + 1

# 3. Filtrar el diccionario para quedarnos solo con las palabras que aparecen más de 1 vez
palabras_repetidas = {palabra: frecuencia for palabra, frecuencia in conteo_palabras.items() if frecuencia > 1}

# Mostramos los resultados por pantalla
print("--- 1. Set de palabras únicas ---")
print(palabras_unicas)

print("\n--- 2. Diccionario de conteo completo ---")
print(conteo_palabras)

print("\n--- 3. Palabras que aparecen más de 1 vez ---")
print(palabras_repetidas)
```

#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

```python
resultado = []

for num in valores:
    # Evaluamos las condiciones
    es_par = (num % 2 == 0)
    es_mult4 = (num % 4 == 0) # Siguiendo la regla del enunciado para múltiplo de 4
    
    # 1. Si cumple ambas condiciones (es par y múltiplo de 4)
    if es_par and es_mult4:
        resultado.append(f"par-mult3:{num}")
    
    # 2. Si solo es múltiplo de 4
    elif es_mult4:
        resultado.append(f"mult3:{num}")
        
    # 3. Si solo es par
    elif es_par:
        resultado.append(f"par:{num}")
        
    # 4. Si no cumple ninguna, se queda como string normal
    else:
        resultado.append(str(num))

# Mostramos la lista final resultante
print("Lista final procesada:", resultado)
```

#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total". Marca un booleano como "aprobado" si el total es mayor a 50. Finalmente, crea un set con todos los nombres de los usuarios aprobados. 

```python
datos = [
    {"id":1,"nombre":"ana","puntos":(10,20,30)},
    {"id":2,"nombre":"luis","puntos":(5,15)},
    {"id":3,"nombre":"marta","puntos":(100,50,25)}
]
```

```python
# Set para almacenar los nombres de los usuarios aprobados
aprobados_set = set()

for usuario in datos:
    # 1. Poner la primera letra de "nombre" en mayúsculas
    usuario["nombre"] = usuario["nombre"].capitalize()
    
    # 2. Calcular la suma total de "puntos" y guardarla en la clave "total"
    total_puntos = sum(usuario["puntos"])
    usuario["total"] = total_puntos
    
    # 3. Marcar un booleano en "aprobado" si el total es mayor a 50
    if total_puntos > 50:
        usuario["aprobado"] = True
        # 4. Añadir el nombre al set de aprobados
        aprobados_set.add(usuario["nombre"])
    else:
        usuario["aprobado"] = False

# Mostramos los resultados por pantalla
print("--- Lista de datos actualizada ---")
for usuario in datos:
    print(usuario)

print("\n--- Set de usuarios aprobados ---")
print(aprobados_set)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
# Creamos el nuevo diccionario vacio
resultado_estudiantes = {}

for id_estudiante, info in estudiantes.items():
    # 1. Convertir el nombre a mayúsculas para usarlo como clave
    nombre_mayus = info["nombre"].upper()
    notas = info["notas"]
    
    # 2. Calcular el promedio de notas (suma total dividida entre la cantidad de notas)
    promedio = sum(notas) / len(notas)
    
    # 3. Marcar True si el promedio es >= 20, de lo contrario False
    aprobado = promedio >= 20
    
    # 4. Crear la tupla con (nota_max, nota_min)
    max_min = (max(notas), min(notas))
    
    # Asignamos el diccionario estructurado a la clave con el nombre en mayúsculas
    resultado_estudiantes[nombre_mayus] = {
        "promedio": promedio,
        "aprobado": aprobado,
        "max_min": max_min
    }

# Mostramos el resultado final por pantalla
for nombre, datos in resultado_estudiantes.items():
    print(f"{nombre}: {datos}")
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json

# 1. Creamos el archivo 'notas_alumnos.txt' con el contenido indicado
contenido_inicial = """Ana,10,20,30
Luis,5,15
Marta,100,50,25"""

with open("notas_alumnos.txt", "w", encoding="utf-8") as f:
    f.write(contenido_inicial)

# 2. Leemos el archivo y procesamos cada línea
resultado = []

with open("notas_alumnos.txt", "r", encoding="utf-8") as f:
    for linea in f:
        # Quitamos espacios en blanco o saltos de línea al inicio/final
        linea_limpia = linea.strip()
        if not linea_limpia:
            continue  # Omitimos líneas vacías si las hubiera
        
        # Separamos por comas
        partes = linea_limpia.split(",")
        nombre = partes[0].capitalize()
        
        # Convertimos las notas a valores numéricos (float)
        notas = [float(nota) for nota in partes[1:]]
        
        # Calculamos el promedio
        promedio = sum(notas) / len(notas)
        
        # Evaluamos si está aprobado
        aprobado = promedio >= 20
        
        # Creamos el diccionario para el usuario
        alumno_dict = {
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": aprobado
        }
        
        resultado.append(alumno_dict)

# 3. Guardamos la lista final en un archivo 'resultado.json'
with open("resultado.json", "w", encoding="utf-8") as f_json:
    json.dump(resultado, f_json, indent=4, ensure_ascii=False)

# Mostramos el resultado por pantalla para verificar
print("--- Proceso completado con éxito ---")
print("Contenido guardado en 'resultado.json':")
print(json.dumps(resultado, indent=4, ensure_ascii=False))
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
import json
import csv

# Diccionario inicial con los datos
datos = {
    "u1": {"id": 1, "nombre": "ana", "roles": ["user", "admin"]},
    "u2": {"id": 2, "nombre": "luis", "roles": ["user"]},
    "u3": {"id": 3, "nombre": "marta", "roles": ["moderator", "user"]}
}

# 1. Crear el archivo 'usuarios.json' con el contenido del diccionario 'datos'
with open("usuarios.json", "w", encoding="utf-8") as f_json:
    json.dump(datos, f_json, indent=4, ensure_ascii=False)

# 2. Abrir el archivo de nuevo y convertir todos los nombres a mayúsculas
with open("usuarios.json", "r", encoding="utf-8") as f_json:
    datos_cargados = json.load(f_json)

for key, usuario in datos_cargados.items():
    usuario["nombre"] = usuario["nombre"].upper()

# 3. Crear el archivo 'usuarios.csv'
with open("usuarios.csv", "w", newline="", encoding="utf-8") as f_csv:
    writer = csv.writer(f_csv)
    
    # Escribir la cabecera
    writer.writerow(["id", "nombre", "roles"])
    
    # Escribir las filas transformando la lista de roles con el separador '|'
    for usuario in datos_cargados.values():
        roles_str = "|".join(usuario["roles"])
        writer.writerow([usuario["id"], usuario["nombre"], roles_str])

print("--- ¡Archivos usuarios.json y usuarios.csv creados con éxito! ---")

# Verificamos el contenido del CSV imprimiéndolo en pantalla
print("\nContenido generado en usuarios.csv:")
with open("usuarios.csv", "r", encoding="utf-8") as f_csv:
    print(f_csv.read())
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
import json
import re
from collections import Counter

# 1. Crear un archivo 'fragmento-libro.txt' de prueba si no existe
texto_prueba = """
La biología computacional y la bioinformática permiten analizar secuencias de ADN.
El análisis de secuencias de ADN es fundamental en la investigación científica.
La ciencia avanza rápidamente con estas herramientas.
"""

with open("fragmento-libro.txt", "w", encoding="utf-8") as f:
    f.write(texto_prueba)

# 2. Leer el archivo y extraer las palabras
with open("fragmento-libro.txt", "r", encoding="utf-8") as f:
    contenido = f.read().lower()

# Extraemos solo palabras (secuencias de letras u otros caracteres alfanuméricos)
palabras = re.findall(r'\b\w+\b', contenido)

# 3. Contar la frecuencia de cada palabra y obtener las 5 más comunes
conteo = Counter(palabras)
top_5_palabras = dict(conteo.most_common(5))

# 4. Guardar los resultados en 'frecuencias.json'
with open("frecuencias.json", "w", encoding="utf-8") as f_json:
    json.dump(top_5_palabras, f_json, indent=4, ensure_ascii=False)

# Mostrar el resultado por pantalla
print("--- 5 palabras más comunes guardadas en 'frecuencias.json' ---")
print(json.dumps(top_5_palabras, indent=4, ensure_ascii=False))
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
import csv

# Datos iniciales que contendrán los tres archivos
datos_iniciales = [
    ["producto", "cantidad"],
    ["A", 10],
    ["B", 5]
]

archivos = ["ventas1.csv", "ventas2.csv", "ventas3.csv"]

# 1. Creamos los tres archivos CSV de ejemplo con la información solicitada
for nombre_archivo in archivos:
    with open(nombre_archivo, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerows(datos_iniciales)

# 2. Leemos los archivos y sumamos las cantidades por producto
totales_productos = {}

for nombre_archivo in archivos:
    with open(nombre_archivo, "r", encoding="utf-8") as f:
        reader = csv.reader(f)
        header = next(reader)  # Saltamos la cabecera ("producto", "cantidad")
        
        for fila in reader:
            producto = fila[0]
            cantidad = int(fila[1])  # Convertimos la cantidad a entero para poder sumar
            
            # Acumulamos si ya existe el producto, o lo creamos si es nuevo
            if producto in totales_productos:
                totales_productos[producto] += cantidad
            else:
                totales_productos[producto] = cantidad

# 3. Guardamos los totales en un único archivo llamado 'total_ventas.csv'
with open("total_ventas.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    # Escribimos la cabecera
    writer.writerow(["producto", "cantidad"])
    
    # Escribimos cada producto con su cantidad sumada
    for producto, cantidad in totales_productos.items():
        writer.writerow([producto, cantidad])

# Mostramos el resultado final por pantalla para verificar
print("--- ¡Archivos combinados con éxito en 'total_ventas.csv'! ---")
print("Contenido resultante:")
with open("total_ventas.csv", "r", encoding="utf-8") as f:
    print(f.read())
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
import json

# 1. Crear el archivo 'alumnos.json' con el contenido indicado
datos_iniciales = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]

with open("alumnos.json", "w", encoding="utf-8") as f:
    json.dump(datos_iniciales, f, indent=4, ensure_ascii=False)

# 2. Leer el archivo y generar el diccionario estructurado
resultado = {}

with open("alumnos.json", "r", encoding="utf-8") as f:
    alumnos_lista = json.load(f)

for alumno in alumnos_lista:
    nombre = alumno["nombre"]
    notas = alumno["notas"]
    
    # Cálculos estadísticos
    promedio = sum(notas) / len(notas)
    max_nota = max(notas)
    min_nota = min(notas)
    aprobado = promedio >= 20  # Criterio estándar del taller
    
    # Asignamos el subdiccionario a la clave con el nombre del alumno
    resultado[nombre] = {
        "promedio": promedio,
        "max": max_nota,
        "min": min_nota,
        "aprobado": aprobado
    }

# Mostramos el resultado final por pantalla
print("--- Diccionario final estructurado ---")
print(json.dumps(resultado, indent=4, ensure_ascii=False))
```
