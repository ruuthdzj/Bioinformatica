#### 3.- Con la cadena mgXDyTw, ponla en mayúsculas, luego en minúsculas. Extrae su longitud y divide la cadena por la letra "g"

```python
cadena="mgXDyTw"
```

```python
# 1. Ponerla en mayúsculas
cadena_mayus = cadena.upper()

# 2. Ponerla en minúsculas
cadena_minus = cadena.lower()

# 3. Extraer su longitud
longitud = len(cadena)

# 4. Dividir la cadena por la letra "g"
# Nota: Como hay una 'g' minúscula y otra 'g' en la mitad (mgXDyTw -> m, XDyTw), 
# usaremos .split('g') o convertiremos a minúsculas/mayúsculas para asegurarnos de capturarlas.
cadena_dividida = cadena.lower().split('g')

# Mostramos los resultados por pantalla
print(f"Cadena original: {cadena}")
print(f"1. En mayúsculas: {cadena_mayus}")
print(f"2. En minúsculas: {cadena_minus}")
print(f"3. Longitud: {longitud}")
print(f"4. Dividida por 'g': {cadena_dividida}")
```

#### 7.- Con la secuencia y la lista siguiente, escribe un algoritmo que cuente el número de veces que aparece un elemento de la lista en la secuencia. Pista: hacer un diccionario para almacenar los resultados puede ser buena idea

```python
dna = "AATGATGAACGAC" 
dinucleotides = ['AA','AT','AG','AC', 
                 'TA','TT','TG','TC', 
                 'GA','GT','GG','GC', 
                 'CA','CT','CG','CT'] 
```

```python
# Creamos un diccionario vacío para guardar los conteos
conteo_dinucleotidos = {}

# Recorremos cada dinucleótido de la lista
for dinu in dinucleotides:
    # Usamos .count() para contar cuántas veces aparece ese dinucleótido en la cadena de ADN
    cantidad = dna.count(dinu)
    
    # Guardamos el resultado en el diccionario
    conteo_dinucleotidos[dinu] = cantidad

# Mostramos el diccionario resultante completo
print("Resultados del conteo de dinucleótidos:")
print(conteo_dinucleotidos)
```

##### ¿Cuántos counts tiene TC?

```python
# Creamos el diccionario con los conteos
conteo_dinucleotidos = {}

for dinu in dinucleotides:
    conteo_dinucleotidos[dinu] = dna.count(dinu)

# Mostramos específicamente cuántas veces aparece 'TC'
print("El dinucleótido 'TC' aparece:", conteo_dinucleotidos['TC'], "veces")

# O si prefieres ver todo el diccionario completo con sus resultados:
print("\nDiccionario completo:")
print(conteo_dinucleotidos)
```

##### ¿Puedes imprimir los dinuleótidos que tengan counts igual a 2?

```python
# 1. Creamos el diccionario con los conteos
conteo_dinucleotidos = {}

for dinu in dinucleotides:
    conteo_dinucleotidos[dinu] = dna.count(dinu)

# 2. Filtramos e imprimimos solo los que tienen un conteo de 2
print("Dinucleótidos que aparecen exactamente 2 veces:")
for dinu, count in conteo_dinucleotidos.items():
    if count == 2:
        print(f"- El dinucleótido '{dinu}' aparece {count} veces.")
```
