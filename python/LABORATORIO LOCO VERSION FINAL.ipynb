```python
#INICIO JUEGO (enunciado)
print ("LABORATORIO LOCO --> Inés y Ruth")
print("---------------------------------------------------------------------------------")
print ("BIENVENIDOS AL LABORATORIO LOCO🥼🔬🧬🧫🧪")
print("Tu misión: adivinar los 3 ingredientes secretos.")
print("NORMAS:")
print("- Tienes solo 5 intentos, cada vez que falles se te restará uno")
print("- Cuidado con las mezclas prohibidas, podrías morir⚠️ CAUTIOOOON ☠️🚨")
print("- En cada intento tienes que seleccionar exactamente 3 ingredientes, respetando las mayúsculas y separados por paréntesis")
print("---------------------------------------------------------------------------------")
print("🧪 Reactivos disponibles en la mesa:")
print("Glucosa, ATP, ADN, Proteína, Enzima, Aminoácido, Reactivo_X, Reactivo_Y")
print("---------------------------------------------------------------------------------")
print("¡MUCHA SUERTE!🍀")
#ingredientes y recetas que al empezar la partida se seleccionará una al azar cada vez
ingredientes = ["Glucosa", "ATP", "ADN", "Proteína", "Enzima", "Aminoácido", "Reactivo_X", "Reactivo_Y"]
receta_1 = ["ADN", "ATP", "Enzima"]
receta_2 = ["Glucosa", "ATP", "Enzima"]
receta_3 = ["Aminoácido", "ATP", "Enzima"]
receta_4 = ["ADN", "Proteína", "ATP"]
receta_5 = ["Glucosa", "Aminoácido", "Proteína"]
recetas = [receta_1, receta_2, receta_3, receta_4, receta_5]#lo ponemos sin "" para que guarde las listas, no como si fueran strings
#cada receta tiéne pistas
pista_receta_1 = ["La receta contiene una molécula relacionada con la información genética.", "Uno de sus componentes participa en la transferencia de energía celular.", "Uno de sus componentes actua como catalizador biolígico.", "la receta no contiene glucosa."]
pista_receta_2 = ["La receta contiene una importante fuente de energía.", "Uno de sus componentes participa en la transferencia de energía celular.", "Uno de sus componentes puede acelerar reacciones biológicas.", "La receta no contiene ADN."]
pista_receta_3 = ["La receta contiene una de las unidades basicas que forman las proteínas.", "Uno de sus componentes esta relacionado con la energia celular.", "Uno de sus componentes actua como catalizador biologico.", "La receta no contiene ADN."]
pista_receta_4 = ["La receta contiene material genetico.", "Uno de los componentes esta formado por cadenas de aminoacidos.", "Uno de los componentes participa en la transferencia de energia celular.", "La receta no contiene glucosa."]
pista_receta_5 = ["La receta contiene una fuente de energía.", "Uno de sus componentes es una unidad basica de las proteinas.", "Otro de sus componentes esta formado por cadenas de aminoacidos.", "La receta no contiene ADN."]
pistas = [pista_receta_1, pista_receta_2, pista_receta_3, pista_receta_4, pista_receta_5]
#mezclas que explotan
mezclas_peligrosas = [["Reactivo_X", "Reactivo_Y"], ["ADN", "Reactivo_X"], ["Proteína", "Reactivo_Y"]]

#hacer que la selección de receta sea aleatoria (con su pista correspondiente)
from random import randint # el randint es para sacar un número al azar entre los que yo diga en este caso (0,4) porque hay 5 recetas y empezamos a contar desde el 0.
def obtener_receta_secreta(recetas):
    numero = randint(0,4) # para seleccionar al azar una de las 5 recetas
    #seleccionar receta y pista correspondiente al nº de receta elegida
    receta_elegida = recetas[numero]
    pista_elegida = pistas[numero]
    return receta_elegida, pista_elegida

#el jugador escribe sus reactivos
def pedir_intento (ingredientes):
    entrada = input("\nElige 3 reactivos, escribelos separados por espacios y respetando las mayusculas.").strip()
    eleccion = entrada.split()
#Qiero comprobar que se ha escrito exactamente 3 reactivos y que los escritos existen entre los que he definido
    if len(eleccion) != 3:
        return None
    for palabra in eleccion:
        if palabra not in ingredientes:
            return None
#Evitar que el jugador ponga elementos repetidos
    if len(set(eleccion)) !=3:
        return None
    return eleccion
    
#Intentos del juego al empezar la partida
intentos_maximos = 5
intentos_realizados = 0
juego_ganado = False

receta_secreta, pista_actual = obtener_receta_secreta(recetas)
#mientras queden intentos y no se haya ganado
while intentos_realizados < intentos_maximos and not juego_ganado:
    print(f"\nINTENTO {intentos_realizados +1} de {intentos_maximos}")
    print(f"Pista del laboratorio:{pista_actual}")
    eleccion_usuario = pedir_intento(ingredientes)
    if eleccion_usuario is None:
        print("Error: Debes introducir 3 reactivos válidos, sin repetir y respetando las mayúsculas")
        continue
    intentos_realizados += 1
    #comprobar si se ha mexclado una compinación que explota
    conjunto_usuario = set(eleccion_usuario)
    explota = False
    for peligro in mezclas_peligrosas:
        if set(peligro).issubset(conjunto_usuario):
            explota =  True
            break
    if explota:
        print ("\n¡BOOM!")
        print ("Has mezclado reactivos incompatibles. ¡La poción ha explotado y has perdido la partida!")
        break
    #comprobamos si se ha acertado la receta secreta
    if sorted(eleccion_usuario) == sorted(receta_secreta):
        print ("\n¡Enhorabuena!")
        print(f"¡Has encontrado la receta secreta: {receta_secreta}! Has ganado la partida.")
        juego_ganado = True
        break
        
    #si no ha explotado ni ganado, dar pista de cuántos ingredientes ha acertado
    aciertos = len(set(eleccion_usuario).intersection(set(receta_secreta)))
    print(f"Resultado de la mezcla: Has acertado {aciertos} de 3 ingredientes.")
if not juego_ganado and intentos_realizados >= intentos_maximos:
    print("\n¡Te has quedado sin intentos!")
    print(f"La receta secreta era: {receta_secreta}")
    print("PERDISTEEE - ¡Suerte para la próxima!")
    
```
