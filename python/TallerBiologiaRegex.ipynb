##### 1.- Abre el fichero dna.txt, guarda una lista del contenido del archivo. Devuelve su longitud y su primer caracter

```python
# 1. Abrimos el archivo dna.txt en modo lectura ('r') y con codificación utf-8
# (Asegúrate de que el archivo esté en la misma carpeta que tu Jupyter Notebook)
with open("dna.txt", "r", encoding="utf-8") as archivo:
    # Leemos todo el contenido del archivo como un único texto (string)
    contenido = archivo.read()

# 2. Guardamos el contenido en forma de lista (por ejemplo, separando cada letra o cada línea)
# Si quieres guardar cada carácter como un elemento de la lista:
lista_contenido = list(contenido)

# 3. Obtenemos su longitud y su primer carácter
longitud = len(lista_contenido)
primer_caracter = lista_contenido[0] if longitud > 0 else "El archivo está vacío"

# Mostramos los resultados por pantalla
print(f"Longitud del contenido: {longitud}")
print(f"Primer carácter: {primer_caracter}")
```

##### 2.- Crea dos archivos. Uno almacena los accession que empiezen por 'a' y el otro el resto

```python
accs = ['ab56', 'bh84', 'hv76', 'ay93', 'ap97', 'bd72']
```

```python
# 1. Filtramos los accessions
empiezan_con_a = []
el_resto = []

for acc in accs:
    # Comprobamos si el accession empieza con la letra 'a' (usando .lower() por seguridad)
    if acc.lower().startswith('a'):
        empiezan_con_a.append(acc)
    else:
        el_resto.append(acc)

# 2. Creamos y guardamos el contenido en el primer archivo (empiezan por 'a')
with open('accessions_con_a.txt', 'w', encoding='utf-8') as archivo_a:
    for acc in empiezan_con_a:
        archivo_a.write(acc + '\n')

# 3. Creamos y guardamos el contenido en el segundo archivo (el resto)
with open('accessions_resto.txt', 'w', encoding='utf-8') as archivo_resto:
    for acc in el_resto:
        archivo_resto.write(acc + '\n')

# Comprobación por pantalla para ver que se han separado bien
print("Empiezan por 'a':", empiezan_con_a)
print("El resto:", el_resto)
print("\n¡Archivos 'accessions_con_a.txt' y 'accessions_resto.txt' creados con éxito!")
```

##### 3.- Imprime solamente los accessions de la lista anterior que empiecen por 'a' y no acaben con '6'. Pista: usa endswith() y startswidth()

```python
# Lista de accessions proporcionada
accs = ['ab56', 'bh84', 'hv76', 'ay93', 'ap97', 'bd72']

# Recorremos la lista y filtramos con las condiciones solicitadas
print("Accessions que empiezan por 'a' y NO acaban en '6':")

for acc in accs:
    # startswith('a') comprueba que empiece por 'a'
    # endswith('6') comprueba si acaba en '6' (el operador 'not' invierte la condición para que no acabe en '6')
    if acc.startswith('a') and not acc.endswith('6'):
        print(acc)
```

##### 4.- Escribe una función que cuente el porcentaje de A de una secuencia

```python
def porcentaje_a(secuencia):
    # Verificamos que la secuencia no esté vacía para evitar una división por cero
    if len(secuencia) == 0:
        return 0.0
    
    # Convertimos la secuencia a mayúsculas para contar tanto 'A' como 'a' por igual
    secuencia_mayus = secuencia.upper()
    
    # Contamos cuántas 'A' hay
    cantidad_a = secuencia_mayus.count('A')
    
    # Calculamos el porcentaje
    porcentaje = (cantidad_a / len(secuencia)) * 100
    
    return porcentaje

# --- Ejemplos de prueba ---
secuencia_prueba = "AGCTATAG"
print(f"Secuencia: {secuencia_prueba}")
print(f"Porcentaje de 'A': {porcentaje_a(secuencia_prueba)}%")  # 3 'A' de 8 letras = 37.5%
```

##### ¿Puedes hacer que devuelva dos decimales? ¿Y que lidie con mayúsculas y minúsculas?

```python
def porcentaje_a(secuencia):
    # Verificamos que la secuencia no esté vacía para evitar una división por cero
    if len(secuencia) == 0:
        return 0.0
    
    # 1. Lidiar con mayúsculas y minúsculas convirtiendo todo a mayúsculas
    secuencia_mayus = secuencia.upper()
    
    # 2. Contamos cuántas 'A' hay
    cantidad_a = secuencia_mayus.count('A')
    
    # Calculamos el porcentaje
    porcentaje = (cantidad_a / len(secuencia)) * 100
    
    # 3. Devolvemos el resultado redondeado exactamente a dos decimales
    return round(porcentaje, 2)

# --- Ejemplos de prueba con mayúsculas, minúsculas y decimales ---
print(porcentaje_a("agctatag"))  # Funciona con minúsculas -> 37.5%
print(porcentaje_a("AGCtatAG"))  # Funciona mezclado -> 37.5%
print(porcentaje_a("AAT"))       # Da un decimal largo (66.666...) -> Devuelve 66.67
```

##### ¿Y que devuelva el número de decimales que yo quiero?

```python
def porcentaje_a(secuencia, decimales=2):
    # Verificamos que la secuencia no esté vacía para evitar una división por cero
    if len(secuencia) == 0:
        return 0.0
    
    # 1. Lidiar con mayúsculas y minúsculas convirtiendo todo a mayúsculas
    secuencia_mayus = secuencia.upper()
    
    # 2. Contamos cuántas 'A' hay
    cantidad_a = secuencia_mayus.count('A')
    
    # Calculamos el porcentaje
    porcentaje = (cantidad_a / len(secuencia)) * 100
    
    # 3. Redondeamos usando el número de decimales que el usuario indique (por defecto 2)
    return round(porcentaje, decimales)

# --- Ejemplos de prueba eligiendo los decimales ---
print(porcentaje_a("AAT", 4))  # Devuelve con 4 decimales: 66.6667
print(porcentaje_a("AAT", 1))  # Devuelve con 1 decimal: 66.7
print(porcentaje_a("AAT"))     # Si no pones nada, usa 2 decimales por defecto: 66.67
```

##### ¿Qué pasa si no ponemos un argumento? ¿Sabrías hacer una función encapsulada?

```python
def analizar_secuencia(secuencia):
    
    # --- Función encapsulada (interna) ---
    def es_valida(dna):
        # Comprueba que solo contenga letras válidas de ADN
        letras_validas = {'A', 'C', 'G', 'T'}
        return all(letra in letras_validas for letra in dna.upper())
    
    # Usamos la función interna dentro de la principal
    if not es_valida(secuencia):
        return "Error: La secuencia contiene caracteres no válidos."
    
    # Si es válida, calculamos por ejemplo su longitud
    return f"Secuencia válida. Longitud total: {len(secuencia)}"

# --- Pruebas ---
print(analizar_secuencia("ACGTACGT"))  # Devuelve: Secuencia válida...
print(analizar_secuencia("ACGX-123"))  # Devuelve: Error...
```

```python
def porcentaje_a(secuencia, decimales=2):
    if len(secuencia) == 0:
        return 0.0
    secuencia_mayus = secuencia.upper()
    cantidad_a = secuencia_mayus.count('A')
    porcentaje = (cantidad_a / len(secuencia)) * 100
    return round(porcentaje, decimales)

# --- Pruebas unitarias con assert ---

# Prueba 1: Secuencia normal en mayúsculas (3 de 8 = 37.5)
assert porcentaje_a("AGCTATAG") == 37.5, "Error en la prueba 1"

# Prueba 2: Secuencia mezclando minúsculas y mayúsculas
assert porcentaje_a("agctatag") == 37.5, "Error en la prueba 2"

# Prueba 3: Probando con el parámetro personalizado de decimales (2 de 3 = 66.666... -> 66.67)
assert porcentaje_a("AAT", decimales=2) == 66.67, "Error en la prueba 3"

# Prueba 4: Secuencia vacía (debe devolver 0.0)
assert porcentaje_a("") == 0.0, "Error en la prueba 4"

# Si llega hasta aquí sin saltar ningún error, significa que ¡todas las pruebas han pasado con éxito!
print("¡Todas las pruebas con assert se han superado correctamente!")
```
