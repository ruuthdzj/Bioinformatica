### BREAK, CONTINUE, PASS

**Break** sale del bucle más inmediato.
**Continue** vuelve al inicio del bucle más inmediato.
**Pass** no hace niguna acción.

```python
x = 0

while x < 10:
    print('x is currently: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
    if x==3:
        print('x==3')
    else:
        print('continuing...')
        continue
```

```python
x = 0

while x < 10:
    print('x is currently: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
    if x==3:
        print('Breaking because x==3')
        break
    else:
        print('continuing...')
        continue
```

#### 1.- Emplea for, split() e if para crear un código que imprima todas las palabras que empiecen por 's'.

```python
st = 'Print only the words that Start with s in this sentence'
```

```python
lista_palabras = st.split()#convierte de text en lista
#print (lista_palabras)
for palabra in lista_palabras:
    if palabra[0] == "s" or palabra [0] == "S":
        print(palabra)
```

#### 3.- Recorre la siguiente string e imprime  solo las palabras que tengan una longitud par

```python
st = 'Print every word in this sentence that has an even number of letters'
```

```python
# Reemplaza este texto con tu string (cadena) correspondiente
texto = "Aprender programacion en python es muy divertido y util"

# 1. Separamos el texto en una lista de palabras usando los espacios
palabras = texto.split()

# 2. Recorremos cada palabra de la lista
for palabra in palabras:
    # 3. Comprobamos si la longitud de la palabra es par (su resto al dividir entre 2 es 0)
    if len(palabra) % 2 == 0:
        print(palabra)
    
```

#### 4.- Escribe un pequeño algoritmo que imprima los enteros del 1 al 100. 
#### Para múltiplos de 3 escribe 'Fizz' en lugar del número
#### Para múltiplos de 5 escribe 'Buzz'
#### Para los que sean múltiplos de 3 y de 5 escribe 'FizzBuzz'

```python
# Recorremos los números del 1 al 100 (el 101 es para que incluya el 100)
for i in range(1, 101):
    # Primero comprobamos si es múltiplo de 3 Y de 5 a la vez
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    # Si es múltiplo solo de 3
    elif i % 3 == 0:
        print("Fizz")
    # Si es múltiplo solo de 5
    elif i % 5 == 0:
        print("Buzz")
    # Si no es múltiplo de ninguno de los dos, imprimimos el número normal
    else:
        print(i)
```
