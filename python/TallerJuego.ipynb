#### Escribe un programa que coja un número entero del 1 al 100, y los jugadores tienen que tratar de adivinar el número.

#### Las reglas son las siguientes:

1. Si un jugador escoje un número menor que 1 o mayor que 100, el programa dice "FUERA DE LÍMITES"
2. En el primer turno del jugador, si está en un rango de 10 del número dice "¡CALIENTE!". Si está fuera de ese rango dice "¡FRIO!"
3. En todas las rondas siguientes, si el jugador se acerca más que la vez anterior, dice "¡MAS CALIENTE!". Si está más lejos que la anterior vez dice "¡MAS FRIO!"
4. Cuando el jugador acierte el número, les dirá que han acertado y cuántos intentos les ha llevado

```python
import random

# Asegúrate de definir el número oculto si no lo tenías antes arriba
numero_oculto = random.randint(1, 100)
lista_intentos = []

print("Introduce un número entre el 1 y el 100:")

while True:
    try:
        intento = int(input("Introduce un número entre el 1 y el 100: "))
    except ValueError:
        print("Por favor, introduce un número entero válido.")
        continue
        
    # Regla 1: Fuera de límites
    if intento > 100 or intento < 1:
        print("fuera de limites")
    else:
        # Primer turno (cuando la lista aún está vacía)
        if len(lista_intentos) == 0:
            if intento == numero_oculto:
                print(f"¡Acertaste en el primer intento!")
                break
            elif abs(intento - numero_oculto) <= 10:
                print("caliente")
            else:
                print("frio")
        else:
            # Siguientes turnos: comprobar si ha acertado
            if intento == numero_oculto:
                print(f"¡Acertaste con {len(lista_intentos) + 1} intentos!")
                break
            
            # Comparamos la distancia actual con el último intento guardado
            distancia_actual = abs(numero_oculto - intento)
            distancia_anterior = abs(numero_oculto - lista_intentos[-1])
            
            if distancia_actual < distancia_anterior:
                print("mas caliente")
            else:
                print("mas frio")
                
        # Guardamos el intento válido en la lista para las siguientes comparaciones
        lista_intentos.append(intento)
    
```
